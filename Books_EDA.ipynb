### Book - Rating EDA

```python
ratings = pd.read_csv('data/Book-Ratings.csv', sep=";", on_bad_lines="skip", encoding='latin-1')
```

```python
ratings.head()
```

```python
ratings.shape
```

```python
ratings.rename(columns={"User-ID":'user_id' , 'Book-Rating':'rating'} , inplace=True)
ratings.head()
```

```python
Rating_count = ratings['user_id'].value_counts()
Rating_count
```

#### Users with more than 200 reviews

```python
active_user_YorN = Rating_count > 200
active_user_YorN
```

```python
active_user_YorN[active_user_YorN].shape
```

#### Index of Users 

```python
active_user_ids = active_user_YorN[active_user_YorN].index
active_user_ids
```

```python
ratings = ratings[ratings['user_id'].isin(active_user_ids)]
ratings.head()
```

#### Joining Ratings with Books

```python
book_rating = ratings.merge(books , on = 'ISBN')
book_rating.head()
```

```python
book_rating_counts = book_rating.groupby('Book-Title')['rating'].count().reset_index()
book_rating_counts.head()
```

```python
book_rating_counts.rename(columns={'rating':'rating_count'},inplace=True)
book_rating_counts.head()
```

```python
final_df = book_rating.merge(book_rating_counts, on='Book-Title')
final_df.head()
```

```python
final_df = final_df[final_df['rating_count'] > 50]
final_df
```

```python
final_df.drop_duplicates(['user_id' , 'Book-Title'] , inplace = True)
final_df.isnull().sum()
```

```python
ratings.shape
```

### PIVOT Table

```python
book_pivot = final_df.pivot_table(columns= 'user_id' , index = 'Book-Title' , values = 'rating')
book_pivot
```

```python
book_pivot.fillna(0 , inplace = True)
```

### SPARSE Matrix

```python
book_sparse = csr_matrix(book_pivot)
book_sparse
```

### KNN

```python
model = NearestNeighbors(algorithm = 'brute')
```

```python
model.fit(book_sparse)
```

```python
name = book_pivot.index[237]
name
```

```python
book_pivot.iloc[237, :]
```

```python
distance, suggestion = model.kneighbors(book_pivot.iloc[237,:].values.reshape(1, -1) , n_neighbors = 5)
```

```python
distance
```

```python
suggestion
```

## Testing Model

```python
def recommend_book(book_name):
    book_id = np.where(book_pivot.index == book_name)[0][0]
    distance, suggestion = model.kneighbors(book_pivot.iloc[book_id , :].values.reshape(1 , -1), n_neighbors = 5 )

    for i in range(len(suggestion)):
        books = book_pivot.index[suggestion[i]]
        for j in books:
            if j == book_name:
                print(f"You searched '{book_name}'\n")
                print("The suggestion books are: \n")
            else:
                print(j)
```

```python
book_name = "2nd Chance"
recommend_book(book_name)
```
